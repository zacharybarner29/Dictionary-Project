```python
#Functions

def dict_contents():
  for position, players in bears.items():
    print(f'\n{position}')
    for player, info in players.items():
      print(f'  {player}')
      for stat, value in info.items():
        print(f'    {stat.title()}: {value}')

def add_player():
    #This function was modified by ChatGPT and the if statments were generated by ChatGPT: https://chatgpt.com/uc/6ada2a2b-3a1c-83ea-9102-713c18eb907a$0
    while True:
      player = input("What player do you want to add?\n").title() #Capitalizes the first letter of each word
      position = input("What position is that player? (Quarterbacks, Running Backs, Wide Receivers, Tight Ends, Defensive Line, Linebackers, Cornerbacks, Safeties)\n").title()
      if position not in bears:
        print("Sorry. That position does not exist.")
      elif player in bears[position]:
        print('Sorry. That player already exists in that position.')
      else:
        break

    #Creates a dictionary for the new player's stats
    stats = {}
    #Asks for stats based on the player's position
    if position == "Quarterbacks":
        stats["games played"] = int(input("Games played: "))
        stats["total passing yards"] = int(input("Total passing yards: "))
        stats["average passing yards per game"] = float(input("Average passing yards per game: ")) #These stats include decimals
        stats["total passing touchdowns"] = int(input("Total passing touchdowns: "))
        stats["total interceptions"] = int(input("Total interceptions: "))
        stats["total completions"] = int(input("Total completions: "))
        stats["total passing attempts"] = int(input("Total passing attempts: "))
        stats["total rushing yards"] = int(input("Total rushing yards: "))
        stats["total rushing touchdowns"] = int(input("Total rushing touchdowns: "))
    elif position == "Running Backs":
        stats["games played"] = int(input("Games played: "))
        stats["total rushing yards"] = int(input("Total rushing yards: "))
        stats["average rushing yards per game"] = float(input("Average rushing yards per game: "))
        stats["total rushing attempts"] = int(input("Total rushing attempts: "))
        stats["total rushing touchdowns"] = int(input("Total rushing touchdowns: "))
        stats["total receptions"] = int(input("Total receptions: "))
        stats["total receiving yards"] = int(input("Total receiving yards: "))
        stats["total receiving touchdowns"] = int(input("Total receiving touchdowns: "))
    elif position == "Wide Receivers" or position == "Tight Ends":
        stats["games played"] = int(input("Games played: "))
        stats["total receptions"] = int(input("Total receptions: "))
        stats["total receiving yards"] = int(input("Total receiving yards: "))
        stats["average receiving yards per game"] = float(input("Average receiving yards per game: "))
        stats["total receiving touchdowns"] = int(input("Total receiving touchdowns: "))
        if position == "Wide Receivers":
            stats["total rushing yards"] = int(input("Total rushing yards: "))
            stats["total rushing touchdowns"] = int(input("Total rushing touchdowns: "))
    elif position == "Defensive Line":
        stats["games played"] = int(input("Games played: "))
        stats["total tackles"] = int(input("Total tackles: "))
        stats["average tackles per game"] = float(input("Average tackles per game: "))
        stats["total sacks"] = float(input("Total sacks: "))
        stats["total tackles for loss"] = int(input("Total tackles for loss: "))
        stats["total forced fumbles"] = int(input("Total forced fumbles: "))
        stats["total fumble recoveries"] = int(input("Total fumble recoveries: "))
    elif position == "Linebackers":
        stats["games played"] = int(input("Games played: "))
        stats["total tackles"] = int(input("Total tackles: "))
        stats["average tackles per game"] = float(input("Average tackles per game: "))
        stats["total sacks"] = float(input("Total sacks: "))
        stats["total interceptions"] = int(input("Total interceptions: "))
        stats["total passes defended"] = int(input("Total passes defended: "))
        stats["total forced fumbles"] = int(input("Total forced fumbles: "))
        stats["total fumble recoveries"] = int(input("Total fumble recoveries: "))
    elif position == "Cornerbacks" or position == "Safeties":
        stats["games played"] = int(input("Games played: "))
        stats["total tackles"] = int(input("Total tackles: "))
        stats["average tackles per game"] = float(input("Average tackles per game: "))
        stats["total sacks"] = float(input("Total sacks: "))
        stats["total interceptions"] = int(input("Total interceptions: "))
        stats["total passes defended"] = int(input("Total passes defended: "))
        stats["total forced fumbles"] = int(input("Total forced fumbles: "))
        stats["total fumble recoveries"] = int(input("Total fumble recoveries: "))
    bears[position][player] = stats
    print(f'{player} and his stats has been added to {position.lower()}')
    dict_contents()

def remove_player():
  while True:
    player = input('What player do you want to remove?\n').title()
    position = input('What position does he play? (Quarterbacks, Running Backs, Wide Receivers, Tight Ends, Defensive Line, Linebackers, Cornerbacks, Safeties)\n').title()
    if position in bears and player in bears[position]:
      break
    else:
      print('Sorry. That player does not exist in that position')
  del bears[position][player] #Deletes a key value pair from the nested dictionary (I found this on W3 Schools)
  dict_contents()

def compare_stats():
  while True:
    position = input('What position are the players you want to compare? (Quarterbacks, Running Backs, Wide Receivers, Tight Ends, Defensive Line, Linebackers, Cornerbacks, Safeties)\n').title()
    player1 = input('What is the first player you want to compare?\n').title()
    player2 = input('What is the second player you want to compare?\n').title()
    stat = input('What statistic would you like to compare?\n').lower() #Makes the word lowercased (opposite of .title())
    if position not in bears or player1 not in bears[position] or player2 not in bears[position] or stat not in bears[position][player1] or stat not in bears[position][player2]:
      print('Sorry. Please enter valid position, players, and statistic.')
    else:
      break
  player1_stat = bears[position][player1][stat]
  player2_stat = bears[position][player2][stat]
  difference = player2_stat - player1_stat #Calculates difference through subtraction
  percent_change = ((difference)/player1_stat)*100 #Calculates the percent change between both stats
  if difference < 0: #Grammer conditional
    more_less_grammer = 'less'
  else:
    more_less_grammer = 'more'
  print(f"\n{player1}'s {stat}: {player1_stat} \n{player2}'s {stat}: {player2_stat} \n{player2}'s {stat} is {abs(difference)} {more_less_grammer} than {player1}'s ({abs(percent_change)}% {more_less_grammer})")

def greatest_stat():
  while True:
    max_stat = 0 #Starts the greatest stat at 0
    position = input('What position do you want to find the greates statistic of? (Quarterbacks, Running Backs, Wide Receivers, Tight Ends, Defensive Line, Linebackers, Cornerbacks, Safeties)\n').title()
    stat = input('What statistic do you want to find?\n').lower()
    if position not in bears:
      print('Sorry. That position does not exist.')
    else:
      break
  #Checks to see if current stat is great than the previously biggest stat
  for player, info in bears[position].items():
    if info[stat] > max_stat: #">" determines if it is greater than or less than
      max_stat = info[stat]
      max_player = player
  print(f'The most {stat} within the {position.lower()} is {max_stat} by {max_player}')


def player_score():
  total = 0 #Start total at 0
  while True:
    position = input("What position is the player? (Quarterbacks, Running Backs, Wide Receivers, Tight Ends, Defensive Line, Linebackers, Cornerbacks, Safeties)/n ").title()
    player = input('What is the name of the player?\n').title()
    if position not in bears:
      print('Sorry. That position does not exist.')
    elif player not in bears[position]:
      print('Sorry. That player does not play that position.')
    else:
      break
  #Adds up every stat value under a player and divides it by the number of stats (average)
  for key, value in bears[position][player].items():
    total += value
  print(f"{player}'s player score as a {position.lower()} is {total/len(bears[position][player])}")

def stat_update(): #Updates any stat (no calculation)
    while True:
      position = input("What position is the player? (Quarterbacks, Running Backs, Wide Receivers, Tight Ends, Defensive Line, Linebackers, Cornerbacks, Safeties)\n").title()
      if position not in bears:
        print('Sorry. That position does not exist.')
      else:
        break
    while True:
      player = input('What is the name of the player?\n').title()
      if player not in bears[position]:
        print('Sorry. That player does not play that position.')
      else:
        break
    print(f'Stats for {position}:')
    #I used ChatGPT to generate these if statements https://chatgpt.com/s/t_6adfb58e7aa481919b2f2588337d757d$0 (Updated version: https://chatgpt.com/s/t_6ae113d8c3148191b866e2150fa8b78f$0)
    if position == "Quarterbacks":
        print("1. Games Played")
        print("2. Total Passing Yards")
        print("3. Average Passing Yards Per Game")
        print("4. Total Passing Touchdowns")
        print("5. Total Interceptions")
        print("6. Total Completions")
        print("7. Total Passing Attempts")
        print("8. Total Rushing Yards")
        print("9. Total Rushing Touchdowns")
    elif position == "Running Backs":
        print("1. Games Played")
        print("2. Total Rushing Yards")
        print("3. Average Rushing Yards Per Game")
        print("4. Total Rushing Attempts")
        print("5. Total Rushing Touchdowns")
        print("6. Total Receptions")
        print("7. Total Receiving Yards")
        print("8. Total Receiving Touchdowns")
    elif position == "Wide Receivers":
        print("1. Games Played")
        print("2. Total Receptions")
        print("3. Total Receiving Yards")
        print("4. Average Receiving Yards Per Game")
        print("5. Total Receiving Touchdowns")
        print("6. Total Rushing Yards")
        print("7. Total Rushing Touchdowns")
    elif position == "Tight Ends":
        print("1. Games Played")
        print("2. Total Receptions")
        print("3. Total Receiving Yards")
        print("4. Average Receiving Yards Per Game")
        print("5. Total Receiving Touchdowns")
    elif position == "Defensive Line":
        print("1. Games Played")
        print("2. Total Tackles")
        print("3. Average Tackles Per Game")
        print("4. Total Sacks")
        print("5. Total Tackles For Loss")
        print("6. Total Forced Fumbles")
        print("7. Total Fumble Recoveries")
    elif position == "Linebackers":
        print("1. Games Played")
        print("2. Total Tackles")
        print("3. Average Tackles Per Game")
        print("4. Total Sacks")
        print("5. Total Interceptions")
        print("6. Total Passes Defended")
        print("7. Total Forced Fumbles")
        print("8. Total Fumble Recoveries")
    elif position == "Cornerbacks":
        print("1. Games Played")
        print("2. Total Tackles")
        print("3. Average Tackles Per Game")
        print("4. Total Sacks")
        print("5. Total Interceptions")
        print("6. Total Passes Defended")
        print("7. Total Forced Fumbles")
        print("8. Total Fumble Recoveries")
    elif position == "Safeties":
        print("1. Games Played")
        print("2. Total Tackles")
        print("3. Average Tackles Per Game")
        print("4. Total Sacks")
        print("5. Total Interceptions")
        print("6. Total Passes Defended")
        print("7. Total Forced Fumbles")
        print("8. Total Fumble Recoveries")
    while True:
      stat = input("Enter the exact name of the statistic you want to change.\n").lower()
      if stat not in bears[position][player]:
        print(f'Sorry. That statistic does not exist under {position.lower()}.')
      else:
        break
    while True:
      try:
        new_stat = float(input('What is the updated statistic?\n')) #User's input could be whole number or decimal
        break
      except ValueError:
        print('Sorry. Please enter a number.')
    #Sets new stat as to what the user inputted
    bears[position][player][stat] = new_stat
    dict_contents()

#----------------------

#Chat-bot Interface
print('Hello. Welcome to the Bears Dictionary!')
while True:
  while True:
    try:
      print('\nMenu options:')
      option = int(input('1: All positions, players, and stats \n2: Add a player and their statistics \n3: Remove a player \n4: Compare statistics between two players \n5: Find the greatest statistic in a position and its player \n6: Find the player score (average of total plays stats) \n7: Update a statistic \n8: EXIT\n\n'))
      break
    except ValueError:
      print('Sorry. Please enter a number only.')
  if option == 1:
    dict_contents()
  elif option == 2:
    add_player()
  elif option == 3:
    remove_player()
  elif option == 4:
    compare_stats()
  elif option == 5:
    greatest_stat()
  elif option == 6:
    player_score()
  elif option == 7:
    stat_update()
  elif option == 8:
    break
  else:
    print('Sorry. Please only enter a number between 1 and 8')
```
